```python
from datasets import load_dataset
import pandas as pd

MMMLU_CONFIGS = {
    "spanish": "ES_LA",
    "hindi": "HI_IN",
    "italian": "IT_IT",
    "chinese": "ZH_CN",
}

for language, config in MMMLU_CONFIGS.items():
    print("=" * 80)
    print(f"{language.upper()} / {config}")
    print("=" * 80)

    ds = load_dataset("openai/MMMLU", config, split="test")
    df = pd.DataFrame(ds)

    print("Number of examples:", len(df))
    print("Columns:", df.columns.tolist())

    print("\nNumber of subjects:", df["Subject"].nunique())

    print("\nSubject distribution:")
    print(df["Subject"].value_counts(dropna=False).sort_index())

    print("\nSubject distribution, sorted by size:")
    print(df["Subject"].value_counts(dropna=False).sort_values(ascending=False))

    print("\nSmallest subjects:")
    print(df["Subject"].value_counts(dropna=False).sort_values().head(10))

    print("\nLargest subjects:")
    print(df["Subject"].value_counts(dropna=False).sort_values(ascending=False).head(10))

    print("\n")
```
